# 01 Explore the dataset
Counts per split and farm label, image sizes, a contact sheet. Set `CONFIG` to `configs/real_data.yaml` for real data.

```python
import matplotlib.pyplot as plt
import numpy as np
from exw.config import load_config
from exw.data.dataset import load_image
from exw.data.manifest import load_manifest

CONFIG = 'configs/default.yaml'
cfg = load_config(CONFIG)
df = load_manifest(cfg['paths']['data_dir'])
df.groupby(['split', 'farm_label']).size().unstack(fill_value=0)
```

```python
# sites per split: no site may appear in two splits
print(df.groupby('split')['site_id'].nunique())
assert (df.groupby('site_id')['split'].nunique() == 1).all()
```

```python
# contact sheet: two images per farm label from the training split
tr = df[df.split == 'train']
picks = [tr[tr.farm_label == lab].head(2) for lab in ['exw_suspect', 'other_condition', 'healthy']]
rows = [r for p in picks for _, r in p.iterrows()]
fig, ax = plt.subplots(3, 2, figsize=(10, 11))
for a, r in zip(ax.ravel(), rows):
    a.imshow(load_image(cfg['paths']['data_dir'], r['path'])); a.set_title(f"{r['image_id']} ({r['farm_label']})", fontsize=8); a.axis('off')
plt.tight_layout()
```
